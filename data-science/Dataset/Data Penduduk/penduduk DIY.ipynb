```python
import pandas as pd

# BACA FILE EXCEL

laki = pd.read_excel('/content/Penduduk Laki Laki Yogyakarta.xlsx', header=1)
perempuan = pd.read_excel('/content/Penduduk Perempuan Yogyakarta.xlsx', header=1)

# RENAME KOLOM

laki.columns = [
    'Kelompok_Umur',
    2021,
    2022,
    2023,
    2024,
    2025,
    2026
]

perempuan.columns = [
    'Kelompok_Umur',
    2021,
    2022,
    2023,
    2024,
    2025,
    2026
]
# HAPUS BARIS KOSONG

laki = laki.dropna(how='all')
perempuan = perempuan.dropna(how='all')

# HAPUS BARIS JUMLAH & CATATAN

laki = laki[
    ~laki['Kelompok_Umur'].astype(str).str.contains(
        'Jumlah|Catatan',
        case=False,
        na=False
    )
]

perempuan = perempuan[
    ~perempuan['Kelompok_Umur'].astype(str).str.contains(
        'Jumlah|Catatan',
        case=False,
        na=False
    )
]

# UBAH DATA JADI NUMERIK

kolom_tahun = [2021, 2022, 2023, 2024, 2025, 2026]

for col in kolom_tahun:
    laki[col] = pd.to_numeric(laki[col], errors='coerce')
    perempuan[col] = pd.to_numeric(perempuan[col], errors='coerce')

# TAMBAH KOLOM GENDER

laki['Gender'] = 'Laki-laki'
perempuan['Gender'] = 'Perempuan'

# GABUNG DATA

df = pd.concat([laki, perempuan], ignore_index=True)

# CEK DATA

print(df.head())

print("\nMissing Value:")
print(df.isnull().sum())

print("\nJumlah Duplikat:")
print(df.duplicated().sum())

# HITUNG TOTAL PENDUDUK

total_laki = laki[kolom_tahun].sum()
total_perempuan = perempuan[kolom_tahun].sum()

# TABEL TOTAL

total_penduduk = pd.DataFrame({
    'tahun': kolom_tahun,
    'laki_laki': total_laki.values,
    'perempuan': total_perempuan.values
})

print("\nTotal Penduduk:")
print(total_penduduk)

# 11. SIMPAN HASIL

df.to_excel('Data_Gabungan_Yogyakarta.xlsx', index=False)

total_penduduk.to_excel('Total_Penduduk_Yogyakarta.xlsx', index=False)

print("\nData berhasil disimpan")
```
