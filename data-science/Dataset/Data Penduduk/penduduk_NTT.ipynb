```python
import pandas as pd
```

```python
df = pd.read_excel(
    "Jumlah Penduduk NTT.xlsx",
    sheet_name="Data 1",
    header=[1, 2]
)
display(df)
print("=="*20)
display(df.info())



```

```python
df.drop(index=[22], inplace=True)
display(df)
```

```python
df_umur = df.iloc[0:22]
total_laki = df_umur.iloc[:, 1:7].sum().values
total_perempuan = df_umur.iloc[:, 7:13].sum().values
df_ringkasan = pd.DataFrame({
    'Tahun': ['2021', '2022', '2023', '2024', '2025', '2026'],
    'Laki-Laki': total_laki,
    'Perempuan': total_perempuan
})

display(df_ringkasan)
df_ringkasan.to_csv('Jumlah Penduduk NTT.csv', index=False)
```
